## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric will be **Precision@K**. This measures how many of the top K pages in the ranked queue correspond to pages that later show the defined observed opportunity outcome. This metric fits the real decision because a reviewer has limited time and will usually inspect only the highest-priority pages first. A higher Precision@K means the ranking places more useful candidates near the top of the review queue. The value of K will be chosen based on a realistic review capacity rather than selected after seeing the results.


```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# ML-03 Section 3: Define the evaluation metric

evaluation_metric = "Precision@K"

print("Primary success metric:", evaluation_metric)
print("Interpretation: among the top K ranked pages,")
print("how many correspond to the observed future opportunity outcome?")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule could rank pages using one or two manually chosen thresholds, but the content-prioritization problem can involve several signals at the same time, such as impressions, sessions, content age, and observed performance movement. Their relationships may not be captured well by a single if-statement or manually chosen threshold.

ML can be useful if these signals contain a measurable pattern that helps distinguish higher-priority pages from lower-priority pages. I will first establish a transparent rule-based baseline and then compare it with a learned ranking or scoring approach. ML will only be considered useful if it improves the agreed evaluation metric, such as Precision@K, on data that was not used to develop the approach.

The goal is therefore not to replace human judgment, but to provide better decision-support for prioritizing pages for review.



```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

# ML-03 — Section 5: Check that multiple measurable signals are available

signal_columns = [
    "impressions_90d",
    "sessions_90d",
    "content_age_days",
    "trend_direction"
]

available_signals = [
    col for col in signal_columns
    if col in df.columns
]

print("Signals available for future ranking/scoring:")
for signal in available_signals:
    print("-", signal)

print(f"\nNumber of available signals: {len(available_signals)}")
```
